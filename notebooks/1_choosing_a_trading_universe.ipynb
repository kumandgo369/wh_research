```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Choosing a universe

## Parse data into dataframes

```python
PROJECT_ROOT = Path.cwd().parent
data_500k = PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_500k.parquet"
data_1m =  PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_1m.parquet"
data_2m =  PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_2m.parquet"
data_4m =  PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_4m.parquet"
data_1m_500k = PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_1m_exit_500k.parquet"
data_2m_1m = PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_2m_exit_1m.parquet"
data_4m_2m = PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_4m_exit_2m.parquet"
data_8m_4m = PROJECT_ROOT/"data/hyperliquid/universe_1d_min_mdv_8m_exit_4m.parquet"
```

```python
df_500k = pd.read_parquet(data_500k)
df_1m = pd.read_parquet(data_1m)
df_2m = pd.read_parquet(data_2m)
df_4m = pd.read_parquet(data_4m)
df_1m_500k = pd.read_parquet(data_1m_500k)
df_2m_1m = pd.read_parquet(data_2m_1m)
df_4m_2m = pd.read_parquet(data_4m_2m)
df_8m_4m = pd.read_parquet(data_8m_4m)
```

## View number of symbols in each univserse over time

### Equal entry / exit minimum 30d median daily volume

```python
n_symbols_500k = df_500k.groupby("open_time")["symbol"].nunique()

n_symbols_500k.plot(figsize=(12, 5), title="Number of symbols over time (500k min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_1m = df_1m.groupby("open_time")["symbol"].nunique()

n_symbols_1m.plot(figsize=(12, 5), title="Number of symbols over time (1m min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_2m = df_2m.groupby("open_time")["symbol"].nunique()

n_symbols_2m.plot(figsize=(12, 5), title="Number of symbols over time (2m min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_4m = df_4m.groupby("open_time")["symbol"].nunique()

n_symbols_4m.plot(figsize=(12, 5), title="Number of symbols over time (4m min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

### Entry / Exit minimum 30d median daily volume (1/2 ratio)

```python
n_symbols_1m_500k = df_1m_500k.groupby("open_time")["symbol"].nunique()

n_symbols_1m_500k.plot(figsize=(12, 5), title="Number of symbols over time (1m entry 500k exit min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_2m_1m = df_2m_1m.groupby("open_time")["symbol"].nunique()

n_symbols_2m_1m.plot(figsize=(12, 5), title="Number of symbols over time (2m entry 1m exit min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_4m_2m = df_4m_2m.groupby("open_time")["symbol"].nunique()

n_symbols_4m_2m.plot(figsize=(12, 5), title="Number of symbols over time (4m entry 2m exit min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
n_symbols_8m_4m = df_8m_4m.groupby("open_time")["symbol"].nunique()

n_symbols_8m_4m.plot(figsize=(12, 5), title="Number of symbols over time (8m entry 4m exit min mdv universe)")
plt.ylabel("symbols")
plt.xlabel("time")
plt.show()
```

```python
print(f"STD/Mean of universe size (500k min mdv): {n_symbols_500k.std() / n_symbols_500k.mean()}")
print(f"STD/Mean of universe size (1m min mdv): {n_symbols_1m.std() / n_symbols_1m.mean()}")
print(f"STD/Mean of universe size (2m min mdv): {n_symbols_2m.std() / n_symbols_2m.mean()}")
print(f"STD/Mean of universe size (4m min mdv): {n_symbols_4m.std() / n_symbols_4m.mean()}")
print(f"STD/Mean of universe size (1m 500k min mdv): {n_symbols_1m_500k.std() / n_symbols_1m_500k.mean()}")
print(f"STD/Mean of universe size (2m 1m min mdv): {n_symbols_2m_1m.std() / n_symbols_2m_1m.mean()}")
print(f"STD/Mean of universe size (4m 2m min mdv): {n_symbols_4m_2m.std() / n_symbols_4m_2m.mean()}")
print(f"STD/Mean of universe size (8m 4m min mdv): {n_symbols_8m_4m.std() / n_symbols_8m_4m.mean()}")
```
